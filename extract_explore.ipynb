```python
import pandas as pd
import requests

print("All good!")
```

```python
# Products ka data DummyJSON API se nikalna
url_products = "https://dummyjson.com/products?limit=0"
response = requests.get(url_products)

print("Status Code:", response.status_code)

# Data JSON format mein hai, lekin DummyJSON ek dictionary deta hai
# jisme "products" key ke andar asal list hoti hai
data = response.json()
products_data = data["products"]

print("Total products:", len(products_data))

# Pehla product dekho
products_data[0]
```

```python
# Users (customers) ka data nikalna
url_users = "https://dummyjson.com/users?limit=0"
response = requests.get(url_users)

print("Status Code:", response.status_code)

data = response.json()
users_data = data["users"]

print("Total users:", len(users_data))

users_data[0]
```

```python
# Carts (customer orders) ka data nikalna
url_carts = "https://dummyjson.com/carts?limit=0"
response = requests.get(url_carts)

print("Status Code:", response.status_code)

data = response.json()
carts_data = data["carts"]

print("Total carts:", len(carts_data))

carts_data[0]
```

```python
import os

# Raw data ke liye folder banate hain (agar already nahi hai)
os.makedirs("data/raw", exist_ok=True)

print("Folder ready!")
```

```python
import json

# Products raw data save karo
with open("data/raw/products_raw.json", "w") as f:
    json.dump(products_data, f, indent=2)

# Users raw data save karo
with open("data/raw/users_raw.json", "w") as f:
    json.dump(users_data, f, indent=2)

# Carts raw data save karo
with open("data/raw/carts_raw.json", "w") as f:
    json.dump(carts_data, f, indent=2)

print("Raw data saved successfully!")
```
